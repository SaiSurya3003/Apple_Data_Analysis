```
file_location = "/FileStore/tables/Customer_Updated.csv"
file_type = "csv"

# CSV Options
infer_schema = "true"
first_row_header = "true"
delimeter = ","

# Reading the File
df = spark.read.format(file_type).option("inferSchema", infer_schema).option("header", first_row_header).option("sep", delimeter).load(file_location)

display(df)
df.show()
```

```
# Create a View or Table for temporary purpose to use it in this Notebook only

temp_table_name = "customer_delta_table"
df.createOrReplaceTempView(temp_table_name)
```

```
%sql
select * from customer_delta_table
```

```
permanent_table_name = "customer_delta_table_2"
df.write.format("parquet").saveAsTable(permanent_table_name)
```
